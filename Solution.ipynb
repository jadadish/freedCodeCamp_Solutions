<h1>Narcissistic Number<h1>
<h3>Given a positive integer, determine whether it is a narcissistic number.<h3>

<h5><i>A number is narcissistic if the sum of each of its digits raised to the power of the total number of digits equals the number itself.<br>
For example, 153 has 3 digits, and 13 + 53 + 33 = 153, so it is narcissistic.
</i><h5>

```python
def is_narcissistic(n):
    
    if not isinstance(n, int):
        return "n should be integer"


    temp = str(n) # converting the int in str
    str_len = len(temp)
    total = 0

    for i in temp:
        total += int(i) ** str_len

    if total == n:
        return True

    return False

print(is_narcissistic(153))     # True
print(is_narcissistic(154))     # False
print(is_narcissistic(371))     # True
print(is_narcissistic(512))     # False
print(is_narcissistic(9))       # True
print(is_narcissistic(11))      # False
print(is_narcissistic(9474))    # True
print(is_narcissistic(6549))    # False
```

<h1>ISBN-13 Validator<h1>
<h3>
Given a string, determine if it is a valid ISBN-13 number.
<h3>
<h5><i>
A valid ISBN-13:
<br>
Contains only digits and hyphens<br>
Has exactly 13 digits after removing hyphens<br>
Passes the following check:<br>
Multiply each digit by 1 or 3, alternating (multiply the first digit by 1, the second by 3, the third by 1, and so on).<br>
The sum of the results must be divisible by 10.
</i><h5>

```python
def is_valid_isbn_13(s):

    temp = ''
    total = 0

    for i in s:
        if i.isdigit():
            temp += i
    
    for i in range(len(temp)):
        if i%2 == 0:
            total += int(temp[i]) * 1
        else:
            total += int(temp[i]) * 3
            
    if len(temp) == 13 and total % 10 == 0:
        return True

    return False

print(is_valid_isbn_13("9780306406157"))                # True
print(is_valid_isbn_13("97803064061570"))               # False
print(is_valid_isbn_13("978-0-13-595705-9"))            # True
print(is_valid_isbn_13("978-030-64061A-4"))             # False
print(is_valid_isbn_13("9-7-8-0-1-3-4-7-5-7-5-9-9"))    # True
```

<h1>Character Frequency<h1>
<h3>
Given a string, return an object (JavaScript) or dictionary (Python) mapping each character to the number of times it appears.
<h3>

```python
def get_frequency(s):

    temp = {}

    a = list(set(s))

    for i in range(len(a)):
        temp[a[i]] = s.count(a[i])

    return temp

print(get_frequency('test'))                                            # {'e': 1, 's': 1, 't': 2}
print(get_frequency('mississippi'))                                     # {'s': 4, 'p': 2, 'm': 1, 'i': 4}
print(get_frequency('hello world'))                                     # {'e': 1, 'r': 1, ' ': 1, 'd': 1, 'h': 1, 'l': 3, 'w': 1, 'o': 2}
print(get_frequency("She sells seashells by the seashore."))            # {'e': 7, 'r': 1, '.': 1, ' ': 5, 'h': 4, 's': 7, 'a': 2, 'l': 4, 'y': 1, 'S': 1, 't': 1, 'b': 1, 'o': 1}
print(get_frequency("The quick brown fox jumps over the lazy dog."))    # {'r': 2, ' ': 8, 's': 1, 'q': 1, 'm': 1, 'g': 1, 'e': 3, 'j': 1, 'z': 1, 'd': 1, 'h': 2, 'f': 1, 'T': 1, 'i': 1, 'u': 2, 'c': 1, 'x': 1, 'a': 1, 'k': 1, 'w': 1, 'n': 1, 'p': 1, 'y': 1, 't': 1, 'b': 1, 'v': 1, '.': 1, 'l': 1, 'o': 4}
```
