<h1>Narcissistic Number<h1>
<h3>Given a positive integer, determine whether it is a narcissistic number.<h3>

<h5><i>A number is narcissistic if the sum of each of its digits raised to the power of the total number of digits equals the number itself.<br>
For example, 153 has 3 digits, and 13 + 53 + 33 = 153, so it is narcissistic.
</i><h5>

```python
def is_narcissistic(n):
    
    if not isinstance(n, int):
        return "n should be integer"


    temp = str(n) # converting the int in str
    str_len = len(temp)
    total = 0

    for i in temp:
        total += int(i) ** str_len

    if total == n:
        return True

    return False

print(is_narcissistic(153))     # True
print(is_narcissistic(154))     # False
print(is_narcissistic(371))     # True
print(is_narcissistic(512))     # False
print(is_narcissistic(9))       # True
print(is_narcissistic(11))      # False
print(is_narcissistic(9474))    # True
print(is_narcissistic(6549))    # False
```
